```python
# Exercise 3: Numbers and Maths
print("I will now count my chickens:") # This line prints I will now count my chickens:

print("Hens", 25 + 30 / 6) # This line prints Hens 30.0
print("Roosters", 100 - 25 * 3 % 4) # This line prints Roosters 97

print("Now I will count the eggs:") # This line prints Now I will count the eggs

print(3 + 2 + 1 - 5 + 4 % 2 - 1 / 4 + 6) # This line prints 6.75

print("Is it true that 3 + 2 < 5 - 7?") # This line prints Is it true that 3 + 2 < 5 - 7?

print(3 + 2 < 5 - 7) # This line prints False

print("What is 3 + 2?", 3 + 2) # This line prints What is 3 + 2? 5
print("What is 5 - 7?", 5 - 7) # This line prints What is 5 - 7? -2

print("Oh, that's why it's False.") # This line prints Oh, that's why it's False.

print("How about some more.") # This line prints How about some more.

print("Is it greater?", 5 > -2) # This line prints Is it greater? True
print("Is it greater or equal?", 5 >= -2) # This line prints Is it greater or equal? True
print("Is it less or equal", 5 <= -2) # This line prints Is it less or equal? False
```
